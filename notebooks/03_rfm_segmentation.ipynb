# 03 — RFM Customer Segmentation

```python
import sys, pandas as pd, plotly.express as px
from sqlalchemy import create_engine
sys.path.append('../src')
from rfm import compute_rfm, segment_summary

engine = create_engine('sqlite:///../data/ecommerce.db')
df = pd.read_sql('SELECT * FROM master', engine, parse_dates=['order_purchase_timestamp'])
rfm = compute_rfm(df)
rfm.head()
```

## Segment distribution

```python
counts = rfm['Segment'].value_counts().reset_index()
counts.columns = ['Segment','Customers']
px.bar(counts, x='Segment', y='Customers', color='Segment', title='Customers per RFM segment').show()
```

## Segment economics

```python
segment_summary(rfm).round(2)
```

## Recency vs Monetary scatter

```python
px.scatter(rfm, x='Recency', y='Monetary', color='Segment', log_y=True, opacity=0.5).show()
```

## Save

```python
rfm.to_csv('../data/processed/rfm_segments.csv', index=False)
print('Saved', len(rfm), 'customers')
```
