# 05 — Churn Prediction

Predict whether a customer will return within the next 180 days using behavioural features.

```python
import sys, pandas as pd
from sqlalchemy import create_engine
sys.path.append('../src')
from churn import build_churn_features, train_churn_model

engine = create_engine('sqlite:///../data/ecommerce.db')
df = pd.read_sql('SELECT * FROM master', engine, parse_dates=['order_purchase_timestamp'])
feats = build_churn_features(df, churn_window_days=180)
feats.head()
```

```python
clf, metrics = train_churn_model(feats)
print('AUC:', round(metrics['auc'], 3))
print('Feature importance:')
metrics['feature_importance']
```

```python
import plotly.express as px
fi = metrics['feature_importance'].reset_index()
fi.columns = ['feature','importance']
px.bar(fi, x='importance', y='feature', orientation='h', title='Churn feature importance').show()
```
